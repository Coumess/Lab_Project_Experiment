```python
import pandas as pd
```

```python
df_word = pd.read_csv("dataset_words.csv",sep=";", decimal=",",encoding="latin-1")
df_word.head()
```

```python
df_word.columns
```

```python
df_word[' MeanVal']=pd.to_numeric(df_word[' MeanVal'],errors = 'coerce')
```

```python
df_word_positive = df_word[df_word[' MeanVal']>=2]
len(df_word_positive)
```

```python
df_word_negatif = df_word[df_word[' MeanVal']<=-2]
len(df_word_negatif)
```

```python
df_word_neutre=df_word[(df_word[' MeanVal'] >= -0.5) & (df_word[' MeanVal'] <= 0.5)]
len(df_word_neutre)
```
