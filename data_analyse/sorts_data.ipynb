## Faces

### Loading of our data

Here we aim for the right sheet (Main Dataset) and create our subset

```python
df = pd.read_excel("dataset.xlsx", sheet_name=1, header=7)
df.head()
```

```python
print(df.columns.tolist())
```

```python
df['Attractive'] = pd.to_numeric(df['Attractive'], errors='coerce')
df_attractivness = df.dropna(subset=['Attractive']) 
df_attractivness = df_attractivness.sort_values("Attractive", ascending=False)
df_attractivness.head()
```

> We create our Dataset of 60 faces, 30 of the most attractive and the less attractive.

```python
df_attractivness = df_attractivness[['Model', 'Attractive', 'GenderSelf']]
df_top = df_attractivness.head(30)
df_bot = df_attractivness.tail(30)
df_final = pd.concat([df_top, df_bot])
df_final
```

## Words

```python
df_word = pd.read_excel("dataset_words.xlsx", nrows=525)
df_word.head()
```

```python
df_word.columns
```

```python
df_word_neutre = df_word[(df_word[' MeanVal'] >= -0.21) & (df_word[' MeanVal'] <= 0.225)]
len(df_word_neutre)
```

```python
df_word_neutre
```

```python

df_word = df_word.sort_values(by=" MeanVal", ascending=True)
df_word_top = df_word.head(20); df_word_bot = df_word.tail(20)
df_word_final = pd.concat([df_word_top, df_word_neutre, df_word_bot])
```

```python
df_word_final = df_word_final[['Trait', ' MeanVal']]
df_word_final
```
